## $\bullet$ Analisando redução média do tamanho dos abstracts 

```python
import pandas as pd 
```

```python
data_pre_filtered = pd.read_csv("resultados/filtered_abstracts_full.csv", low_memory=False)
data = pd.read_csv("resultados/filtered_abstracts_post.xls")

print(len(data_pre_filtered))
print(len(data))

display(data[["Abstract", "Filtered Abstracts"]].head(10))
```

```python
tamanho_original = data["Abstract"].str.split().str.len()
tamanho_filtrado = data["Filtered Abstracts"].str.split().str.len()

razoes = tamanho_filtrado / tamanho_original

razao_media = razoes.mean()

print(f"Razão média: {razao_media:.3f}")
print(f"Percentual médio mantido: {razao_media * 100:.2f}%")
```
